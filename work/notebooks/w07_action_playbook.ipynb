## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

## Ranked Action Queue

| Rank | Action | Reason Code | Priority |
|--------|--------|--------|--------|
| 1 | Refresh high-traffic content older than 6 months | REFRESH_OLD | High |
| 2 | Improve pages with declining engagement | ENGAGEMENT_DROP | High |
| 3 | Update outdated examples and references | CONTENT_STALE | Medium |
| 4 | Expand thin content sections | LOW_DEPTH | Medium |
| 5 | Improve internal linking | LINK_GAPS | Low |

## Reason Codes

REFRESH_OLD:
Content has aged and may no longer reflect current information.

ENGAGEMENT_DROP:
User interaction metrics show declining engagement.

CONTENT_STALE:
References and examples are outdated.

LOW_DEPTH:
Topic coverage is insufficient.

LINK_GAPS:
Internal linking opportunities exist.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## Intended Use

This playbook is designed to support content review and prioritization decisions.

The ranked queue helps identify content that may benefit from refresh, expansion, or optimization.

The output is intended for human review before any action is taken.

---

## Limits

- Recommendations are decision-support only.
- The system does not guarantee performance improvement.
- Content quality still requires human evaluation.
- Recommendations should not be treated as automated publishing instructions.
- Outputs are directional rather than predictive guarantees.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review Required

Before publishing any changes:

- Verify factual accuracy.
- Confirm content relevance.
- Review writing quality.
- Check brand alignment.
- Validate external references.

---

## No-Go List

The following actions should NOT be automated:

- Publishing content directly.
- Modifying legal information.
- Editing medical or financial advice.
- Removing content without review.
- Making strategic business decisions solely from model output.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring

Review outputs monthly.

Track:

- Engagement trends
- Recommendation acceptance rate
- Content freshness
- Review outcomes

---

## Retrain Triggers

Retraining should be considered when:

- Recommendation quality declines.
- New content categories appear.
- Data distribution changes significantly.
- Performance metrics decline consistently.
- Human reviewers reject recommendations frequently.
