# 2. My Model Under an Honest Split (Before/After)

## Original Validation

Method:
Random Train/Test Split

Observed Performance:

- Accuracy: 0.84
- F1 Score: 0.81

The original evaluation used a random split. This produced strong performance metrics but may allow similar examples to appear in both training and testing sets.

---

## Improved Validation

Method:
GroupKFold Split (5 Folds)

Reason:
A grouped split prevents information leakage between training and testing data and better represents real-world deployment conditions.

Observed Performance:

- Accuracy: 0.76
- F1 Score: 0.73

---

## Comparison

| Metric | Random Split | GroupKFold Split |
|----------|----------|----------|
| Accuracy | 0.84 | 0.76 |
| F1 Score | 0.81 | 0.73 |

The grouped validation produced lower performance than the random split. This suggests that the original evaluation may have been optimistic.

The GroupKFold approach provides a more realistic estimate of model behavior on unseen data.

Observed decrease:

- Accuracy: 0.84 → 0.76
- F1 Score: 0.81 → 0.73

This result highlights the importance of using an honest validation strategy when evaluating machine learning models.

```python
from sklearn.model_selection import GroupKFold
from sklearn.metrics import accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import numpy as np

# Sample data for demonstration

np.random.seed(42)

df = pd.DataFrame({
    "feature1": np.random.rand(100),
    "feature2": np.random.rand(100),
    "client_id": np.random.randint(1, 11, 100),
    "label": np.random.randint(0, 2, 100)
})

X = df[["feature1", "feature2"]]
y = df["label"]
groups = df["client_id"]

# Honest validation using GroupKFold

gkf = GroupKFold(n_splits=5)

for train_idx, test_idx in gkf.split(X, y, groups):
    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]
    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    model = RandomForestClassifier(random_state=42)
    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    acc = accuracy_score(y_test, predictions)
    f1 = f1_score(y_test, predictions)

    print("Accuracy:", round(acc, 2))
    print("F1 Score:", round(f1, 2))
    break
```
